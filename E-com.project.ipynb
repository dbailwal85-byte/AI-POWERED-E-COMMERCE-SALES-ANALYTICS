```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Sales"])

plt.title("Sales Boxplot")

plt.show()
```

```python
Q1 = df["Sales"].quantile(0.25)

Q3 = df["Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - (1.5 * IQR)

upper_limit = Q3 + (1.5 * IQR)

print("Lower Limit :", lower_limit)
print("Upper Limit :", upper_limit)
```

```python
outliers = df[
    (df["Sales"] < lower_limit) |
    (df["Sales"] > upper_limit)
]

outliers
```

```python
# 7. Outlier Treatment
```

```python
# Removing outliers
df = df[
    (df["Sales"] >= lower_limit) &
    (df["Sales"] <= upper_limit)
]
```

```python
# Phase 7 — Customer Segmentation
# Data Preparation
# Scaling
# K-Means Clustering
# Elbow Method
# Silhouette Score
# Cluster Visualization
# Business Interpretation
```

```python
# Create Customer Dataset
customer_data = df.groupby("Customer ID").agg({
    "Sales":"sum",
    "Order ID":"nunique",
    "Customer Rating":"mean"
}).reset_index()

customer_data.rename(columns={
    "Sales":"Lifetime Spend",
    "Order ID":"Total Orders",
    "Customer Rating":"Average Rating"
}, inplace=True)
```

```python
# Create Average Order Value
customer_data["Average Order Value"] = (
    customer_data["Lifetime Spend"] /
    customer_data["Total Orders"]
)

```

```python
customer_data.head()
```

```python
# Scaling
```

```python
features = customer_data[[
    "Lifetime Spend",
    "Total Orders",
    "Average Order Value",
    "Average Rating"
]]
```

```python
scaler = StandardScaler()

scaled_features = scaler.fit_transform(features)
```

```python
customer_data.isnull().sum()
```

```python
customer_data["Average Rating"] = customer_data["Average Rating"].fillna(
    customer_data["Average Rating"].median()
)
```

```python
features = customer_data[[
    "Lifetime Spend",
    "Total Orders",
    "Average Order Value",
    "Average Rating"
]]

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_features = scaler.fit_transform(features)
```

```python
import numpy as np

np.isnan(scaled_features).sum()
```

```python
# Elbow Method
inertia = []

for k in range(2,11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(scaled_features)

    inertia.append(model.inertia_)
```

```python
plt.figure(figsize=(8,5))

plt.plot(range(2,11), inertia, marker="o")

plt.title("Elbow Method")

plt.xlabel("Number of Clusters")

plt.ylabel("Inertia")

plt.show()
```

```python
# : Train K-Means Model
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

customer_data["Cluster"] = kmeans.fit_predict(
    scaled_features
)
```

```python
# 7: Silhouette Score
score = silhouette_score(
    scaled_features,
    customer_data["Cluster"]
)

print("Silhouette Score :", round(score,3))
```

```python
# Cluster Visualization
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=customer_data,
    x="Lifetime Spend",
    y="Total Orders",
    hue="Cluster",
    palette="Set2",
    s=100
)

plt.title("Customer Segments")

plt.show()
```

```python
# Analyze Each Cluster
customer_data.groupby("Cluster")[[
    "Lifetime Spend",
    "Total Orders",
    "Average Order Value",
    "Average Rating"
]].mean()
```

```python
# Business Interpretation
cluster_names = {
    0: "Low Value Customers",
    1: "Premium Customers",
    2: "Regular Customers"
}

customer_data["Customer Segment"] = (
    customer_data["Cluster"]
    .map(cluster_names)
)
```
